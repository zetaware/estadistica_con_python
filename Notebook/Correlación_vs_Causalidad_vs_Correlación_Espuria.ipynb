# **ESCENARIO 1: Causalidad (REAL)**
Cigarrillos vs Cáncer

# **Explicación**

A mayor consumo de cigarrillos, mayor probabilidad de cáncer.

* Tiene lógica médica
* Tiene evidencia
* Es causalidad

# **ESCENARIO 2: Correlación Espuria**

# **Ahogamientos y venta de helados**

A simple vista, cuando aumentan las ventas de helado, también aumentan los ahogamientos.

Entonces alguien podría decir:

* “Comprar helados provoca ahogamientos”

Pero eso es falso.

Lo que ocurre realmente es que ambas variables aumentan por una tercera variable:

### **la temperatura**

Cuando hace más calor:

* más personas compran helado
* más personas van a playas, piscinas o ríos
* al haber más personas nadando, aumentan los ahogamientos

Entonces:

* helados y ahogamientos están correlacionados
* pero uno no causa al otro
* la relación está explicada por una variable confusora: temperatura

```python
import numpy as np
import pandas as pd
import plotly.express as px

df = pd.read_csv("/content/dataset_correlacion_espuria.csv")
```

```python
fig = px.scatter(
    df,
    x="ventas_helado",
    y="ahogamientos",
    trendline="ols",
    title="Escenario 2: Ventas de helado vs ahogamientos"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Ventas de helado",
    yaxis_title="Número de ahogamientos"
)

fig.show()
```

# **Escenario 3 — Explicación de la variable oculta**
## **Temperatura vs ventas de helado**
La temperatura explica el comportamiento de ambas variables.
Cuando aumenta la temperatura:
- aumentan las ventas de helado
- aumentan los ahogamientos

Por eso la relación entre helados y ahogamientos no es causal,
sino una correlación espuria causada por una variable confusora: la temperatura.

```python
fig = px.scatter(
    df,
    x="temperatura",
    y="ventas_helado",
    trendline="ols",
    title="Escenario 3A: Temperatura vs ventas de helado"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Temperatura",
    yaxis_title="Ventas de helado"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="temperatura",
    y="ahogamientos",
    trendline="ols",
    title="Escenario 3B: Temperatura vs ahogamientos"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Temperatura",
    yaxis_title="Número de ahogamientos"
)

fig.show()
```
