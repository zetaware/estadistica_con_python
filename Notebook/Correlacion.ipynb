# Explicación del Dataset
El Dataset consiste básicamente en datos sobre dolores de una persona de acuerdo a su altura, sexo, edad y peso.

*   Los datos de la escala de dolor
*   Altura del participante en pulgadas
*   Peso del participante en libras
*   E - Extrovertismo (0-21) en relación con Introvertido, es decir, si la escala de Extrovertido es alta, la escala de Introvertido es baja para esa misma fila de datos

*   I - Introvertismo (0-19) en relación con Extrovertido, es decir, si la escala de Introvertido es alta, la escala de Extrovertido es baja para esa misma fila de datos

*   S - Sensación (0-26) en relación con Intuición, es decir, si la escala de Sensación es alta, la escala de Intuición es baja para esa misma fila de datos

*   N - Intuición (0-21) en relación con Sensación, es decir, si la escala de Intuición es alta, la escala de Sensación es baja para esa misma fila de datos

# Matriz de Correlacion (HeatMap)

```python
# Importar las librerias

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el Dataset

dfDolor = pd.read_csv('/content/drive/MyDrive/Curso Estadistica con Python/DatasetPosturePersonality.csv')

dfDolor.head()
```

```python
#Matriz de Correlacion

corr = dfDolor.corr(numeric_only= True)

corr
```

# Mapa de Calor (HeatMap)

```python
fig, ax  =  plt.subplots(figsize = (15,10))
sns.heatmap(corr, annot = True,cmap = 'YlGnBu', ax = ax, vmax = 1, vmin = -1)
plt.title('Matriz de Correlacion')
plt.show()
```

# Grafico de Dispersion

```python
plt.figure(figsize= (12, 6))
sns.scatterplot(data = dfDolor, x = 'WEIGHT', y  = 'PAIN 4', alpha = 0.7 )

#Configuracion de las etiquetas y el titulo
plt.xlabel('Peso')
plt.ylabel('Dolor de Cuello')
plt.title('Relacion entre el peso y el dolor de cuello', fontsize = 14, fontweight = 'bold')
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize= (12, 6))
sns.scatterplot(data = dfDolor, x = 'E', y  = 'I', alpha = 0.7 )

#Configuracion de las etiquetas y el titulo
plt.xlabel('Extrovertido')
plt.ylabel('Introvertido')
plt.title('Relacion entre el Extrovertismo e Introvertismo', fontsize = 14, fontweight = 'bold')
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize= (12, 6))
sns.scatterplot(data = dfDolor, x = 'HEIGHT', y  = 'E', alpha = 0.7 )

#Configuracion de las etiquetas y el titulo
plt.xlabel('Altura')
plt.ylabel('Extrovertido')
plt.title('Relacion entre la Altura y el Extrovertismo', fontsize = 14, fontweight = 'bold')
plt.grid(True)

plt.show()
```
